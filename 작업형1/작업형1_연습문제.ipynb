```python
# Section 01 필터링, 최솟값, 중앙값
# 1. 'f5' 컬럼이 0이 아닌 데이터(행)를 구하시오.
# 2. 앞에서 구한 데이터에 'views' 컬럼 결측치를 'views' 컬럼의 최솟값으로 채워주세요.
# 3. 그리고 'views' 컬럼의 중앙값을 계산해 정수로 구하시오.

# 기본 코드
import pandas as pd
df = pd.read_csv('type1_data1.csv')
df
```

```python
# Section 05 문자열 슬라이싱, 파생변수, 평균값
# 1. 'f4' 컬럼 데이터에 'FJ'가 포함된 데이터를 찾으시오.
# 2. 찾은 데이터 중에서 'f2' 컬럼의 평균값을 구하시오. (반올림 후 소수 둘째 자리까지 계산)

# 기본 코드
import pandas as pd
df = pd.read_csv('type1_data1.csv')
df
```

```python
# 2. 찾은 데이터 중에서 'f2' 컬럼의 평균값을 구하시오. (반올림 후 소수 둘째 자리까지 계산)
cond = (df['new'] == True)

print(round(df[cond]['f2'].mean(), 2))
```

```python
# Section 06 필터링, 분산
# 1. 'f3' 컬럼이 gold이면서 'f2' 컬럼이 2인 데이터를 찾으시오.
# 2. 찾은 데이터에서 'f1' 컬럼의 분산을 구하시오. (반올림 후 소수 둘째 자리까지 계산)

# 기본 코드
import pandas as pd
df = pd.read_csv('type1_data1.csv')
df
```

```python
# 1. 'f3' 컬럼이 gold이면서 'f2' 컬럼이 2인 데이터를 찾으시오.
cond = ((df['f3'] == 'gold') & (df['f2'] == 2))
df = df[cond]
df
```

```python
# 2. 찾은 데이터에서 'f1' 컬럼의 분산을 구하시오. (반올림 후 소수 둘째 자리까지 계산)
print(round(df['f1'].var(), 2))
```

```python
# Section 07 값 변경(연산), 필터링 절댓값
# 1. 모든 나이(age)에 1을 더하시오.
# 2. 20대의 'views' 평균과 30대의 'views' 평균의 차이의 절댓값을 구하시오.(반올림 후 소수 둘째 자리까지 계산)

# 기본 코드
import pandas as pd
df = pd.read_csv('type1_data1.csv')
df
```

```python
# 1. 모든 나이(age)에 1을 더하시오.
df['age'] = df['age'] + 1
df
```

```python
# 2. 20대의 'views' 평균과 30대의 'views' 평균의 차이의 절댓값을 구하시오.(반올림 후 소수 둘째 자리까지 계산)
cond1 = (df['age'] >= 20) & (df['age'] < 30)
cond2 = (df['age'] >= 30) & (df['age'] < 40)
avg_20 = df[cond1]['views'].mean()
avg_30 = df[cond2]['views'].mean()
avg_abs = avg_20 - avg_30
print(round(abs(avg_abs), 2))
```

```python
# Section 10 그룹핑, 최댓값, 정렬
# 1. 결측치가 있는 행을 삭제하시오.
# 2. 결측치가 삭제된 데이터를 사용하여 지역별(city) 평균을 계산하시오.
# 3. 앞에서 계산한 지역별 평균 데이터에서 'f2' 컬럼 값이 가장 큰 지역을 구하시오.

# 기본 코드
import pandas as pd
df = pd.read_csv('type1_data1.csv')
df
```

```python
# Section 11 슬라이싱, 사분위수, 결측치 제거
# 1. 데이터에서 결측치가 있는 데이터(행)를 모두 제거하시오.
# 2. 결측치가 제거된 데이터를 사용하여 앞에서부터 70% 데이터를 구하시오. (단, 데이터 70% 지점의 index가 소수점으로 계산될 경우 소수점 이하는 버림)
# 3. 앞에서 구한 70% 데이터 중 'views' 컬럼의 3사분위수에서 1사분위수를 뺀 값을 정수로 구하시오.

# 기본 코드
import pandas as pd
df = pd.read_csv('type1_data1.csv')
df
```

```python
# 1. 데이터에서 결측치가 있는 데이터(행)를 모두 제거하시오.
df = df.dropna()
df.info()
```

```python
# 2. 결측치가 제거된 데이터를 사용하여 앞에서부터 70% 데이터를 구하시오. (단, 데이터 70% 지점의 index가 소수점으로 계산될 경우 소수점 이하는 버림)
seventy_percent = round(len(df)*0.7)
df = df.iloc[:seventy_percent]
```

```python
# 3. 앞에서 구한 70% 데이터 중 'views' 컬럼의 3사분위수에서 1사분위수를 뺀 값을 정수로 구하시오.
third = df['views'].quantile(0.75)
one = df['views'].quantile(0.25)
print(int(third - one))
```

```python
# Section 12 결측치 처리, 최빈값, 데이터 개수
# 1. 결측치가 가장 많은 두 컬럼을 찾으시오.
# 2. 첫 번째로 결측치가 많은 컬럼에서 결측치가 있는 데이터(행)를 삭제하시오.
# 3. 두 번째로 결측치가 많은 컬럼을 최빈값으로 대체하시오.
# 4. 'f3' 컬럼의 'gold' 값을 가진 데이터의 수를 정수형으로 구하시오.

# 기본 코드
import pandas as pd
df = pd.read_csv('type1_data1.csv')
df
```

```python
# 1. 결측치가 가장 많은 두 컬럼을 찾으시오.
print(df.isnull().sum()) # 혹은 df.info()로 찾기
```

```python
# 2. 첫 번째로 결측치가 많은 컬럼에서 결측치가 있는 데이터(행)를 삭제하시오.
df = df.dropna(subset = ['f1'])
df.info()
```

```python
# 3. 두 번째로 결측치가 많은 컬럼을 최빈값으로 대체하시오.
mode = df['f3'].mode()[0]
df['f3'] = df['f3'].fillna(mode)
df.info()
```

```python
# 4. 'f3' 컬럼의 'gold' 값을 가진 데이터의 수를 정수형으로 구하시오.
cond = (df['f3'] == 'gold')
print(len(df[cond]))
```
