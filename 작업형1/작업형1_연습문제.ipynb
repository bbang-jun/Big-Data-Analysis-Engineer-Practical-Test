```python
# Section 16 이상치, IQR
# 1. 'views' 컬럼의 1사분위수, 3사분위수 그리고 IQR을 계산하시오.
# 2. 이상치 조건에 맞는 데이터를 찾으시오. (이상치는 1사분위수 - (IQR + 1.5)보다 작은 값과 3사분위수 + (IQR + 1.5)보다 큰 값)
# 3. 이상치 데이터의 'views' 컬럼 합을 정수로 구하시오.

# 기본 코드
import pandas as pd
df = pd.read_csv('type1_data1.csv')
df.info()
```

```python
# 1. 'views' 컬럼의 1사분위수, 3사분위수 그리고 IQR을 계산하시오.
q1 = df['views'].quantile(0.25)
q3 = df['views'].quantile(0.75)
iqr = q3 - q1
print(iqr)
```

```python
# 2. 이상치 조건에 맞는 데이터를 찾으시오. (이상치는 1사분위수 - (IQR * 1.5)보다 작은 값과 3사분위수 + (IQR * 1.5)보다 큰 값)
cond1 = (q1 - 1.5 * iqr)
cond2 = (q3 + 1.5 * iqr)
cond = (df['views'] < cond1) | (df['views'] > cond2)
df = df[cond]
df
```

```python
# Section 22 시간 간의 차이 계산(분), 그룹핑
# 1. 주문 시간과 실제 도착 시간의 차이를 분 단위로 계산하시오.
# 2. 앱 종류별로 평균 도착 시간(분)을 계산하시오.
# 3. 평균적으로 가장 빠른 앱 종류를 찾고, 해당 앱의 평균 도착 시간을 분으로 반올림하여 정수로 구하시오.

# 기본 코드
import pandas as pd
df = pd.read_csv('delivery_time.csv')
df
```

```python
# 1. 주문 시간과 실제 도착 시간의 차이를 분 단위로 계산하시오.
df['주문시간'] = pd.to_datetime(df['주문시간'])
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'])

df['diff'] = (df['실제도착시간'] - df['주문시간']).dt.total_seconds() / 60
df
```

```python
# 1. 각 사용자별로 첫 주문과 마지막 주문 사이의 시간 간격을 일 단위로 계산하시오.
df['주문시간'] = pd.to_datetime(df['주문시간'])
df_first = df.groupby(['user'])['주문시간'].min()
df_last = df.groupby(['user'])['주문시간'].max()
df = (df_last - df_first).dt.days
df
```

```python
# 2. 시간차가 0일인 사용자를 제외하고, 나머지 사용자들의 평균 시간 간격(일 단위)을 계산하시오.
cond = df > 0
df = df[cond]
df
```

```python
# Section 27 시간 범위, 속도(km/h)
# 1. 점심시간(10시부터 13시 전까지)에 주문된 배달 데이터를 찾으시오.
# 2. 점심시간 주문 건 중 과속(평균 속도가 50km/h 이상)하는 주문 수를 정수로 구하시오.
# 배달시간 = 실제도착시간 - 주문시간
# 속도(km/h) = 거리(km) / 시간(h)

# 기본 코드
import pandas as pd
df = pd.read_csv('delivery_time.csv')
df
```

```python
# 1. 점심시간(10시부터 13시 전까지)에 주문된 배달 데이터를 찾으시오.
df['주문시간'] = pd.to_datetime(df['주문시간'])
df['실제도착시간'] = pd.to_datetime(df['실제도착시간'])

df['hour'] = df['주문시간'].dt.hour
cond = (df['hour'] >= 10) & (df['hour'] < 13) # 10시부터 13시 전
df = df[cond]
df
```

```python
# 2. 점심시간 주문 건 중 과속(평균 속도가 50km/h 이상)하는 주문 수를 정수로 구하시오.
# 배달시간 = 실제도착시간 - 주문시간
# 속도(km/h) = 거리(km) / 시간(h)
df['배달시간'] = (df['실제도착시간'] - df['주문시간']).dt.total_seconds() / 60 / 60
df['속도'] = df['거리'] / df['배달시간']
cond = df['속도'] >= 50
print(len(df[cond]))
```
